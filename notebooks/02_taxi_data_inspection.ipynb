```python
jan_start = pd.Timestamp("2025-01-01")
feb_start = pd.Timestamp("2025-02-01")

print(
    "Pickup outside January:",
    ((df["tpep_pickup_datetime"] < jan_start) |
     (df["tpep_pickup_datetime"] >= feb_start)).sum()
)

print(
    "Dropoff outside January:",
    ((df["tpep_dropoff_datetime"] < jan_start) |
     (df["tpep_dropoff_datetime"] >= feb_start)).sum()
)

print(
    "Dropoff before pickup:",
    (df["tpep_dropoff_datetime"] < df["tpep_pickup_datetime"]).sum()
)
```

```python
duration_minutes = (
    df["tpep_dropoff_datetime"] - df["tpep_pickup_datetime"]
).dt.total_seconds() / 60

duration_minutes.describe()
```

```python
print("Negative durations:", (duration_minutes < 0).sum())
print("Over 24 hours:", (duration_minutes > 1440).sum())
```

```python
bad_duration = df[
    df["tpep_dropoff_datetime"] < df["tpep_pickup_datetime"]
][
    [
        "tpep_pickup_datetime",
        "tpep_dropoff_datetime",
        "trip_distance",
        "fare_amount",
        "total_amount",
        "PULocationID",
        "DOLocationID"
    ]
]

bad_duration.head(20)
```

```python
long_trips = df[
    duration_minutes > 1440
][
    [
        "tpep_pickup_datetime",
        "tpep_dropoff_datetime",
        "trip_distance",
        "fare_amount",
        "total_amount"
    ]
]

long_trips.sort_values(
    "tpep_pickup_datetime"
).head(20)
```

```python
# Remove trips with impossible or implausible durations:
# - negative duration
# - duration greater than 24 hours
duration_minutes = (
    df["tpep_dropoff_datetime"] - df["tpep_pickup_datetime"]
).dt.total_seconds() / 60

df_clean = df[
    (duration_minutes >= 0) &
    (duration_minutes <= 1440)
].copy()
```

```python
print("Original rows:", len(df))
print("Clean rows:", len(df_clean))
print("Removed rows:", len(df) - len(df_clean))
```

```python
print(df_clean.shape)
```

```python
clean_path = Path.home() / "datasets" / "nyc-taxi" / "2025" / "yellow_tripdata_2025-01_clean.parquet"

df_clean.to_parquet(clean_path, index=False)
```

```python
print(clean_path)
print(f"Cleaned rows: {len(df_clean):,}")
```

```python
# reviewing missing values found earlier
missing_cols = [
    "Airport_fee",
    "passenger_count",
    "congestion_surcharge",
    "RatecodeID",
    "store_and_fwd_flag"
]

all_missing = df_clean[missing_cols].isna().all(axis=1).sum()
any_missing = df_clean[missing_cols].isna().any(axis=1).sum()

print("All five missing:", all_missing)
print("Any of five missing:", any_missing)
```
